```python
from unsloth import FastLanguageModel
import torch

max_seq_length = 2048
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="Qwen/Qwen2.5-Coder-7B-Instruct", # ya "unsloth/Meta-Llama-3.1-8B-Instruct"
    max_seq_length=max_seq_length,
    load_in_4bit=True, # VRAM bachane ke liye
)

# LoRA adapters attach karein
model = FastLanguageModel.get_peft_model(
    model,
    r=16, # Rank: 8 ya 16 chote dataset ke liye perfect hai
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0, # Unsloth me 0 optimized hai
    bias="none",
)
```

```python
from datasets import load_dataset

# Yeh seedha 300 samples download kar lega:
raw_data = load_dataset("gretelai/synthetic_text_to_sql", split="train[:300]")

prompt_template = """<|im_start|>system
You are a SQL expert. Given the database schema and question, write a valid SQL query.<|im_end|>
<|im_start|>user
Database Schema:
{schema}

Question:
{question}<|im_end|>
<|im_start|>assistant
{sql}<|im_end|>"""

def format_fn(examples):
    texts = []
    for context, q, sql in zip(examples["sql_context"], examples["sql_prompt"], examples["sql"]):
        text = prompt_template.format(schema=context, question=q, sql=sql)
        texts.append(text)
    return {"text": texts}

dataset = raw_data.map(format_fn, batched=True)
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    dataset_num_proc=2,
    packing=False,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        warmup_steps=10,
        max_steps=100, # Chote dataset (200-500 samples) ke liye 60-120 steps kafi hain
        learning_rate=2e-4,
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=10,
        optim="adamw_8bit",
        output_dir="outputs",
    ),
)

trainer.train()
```

```python
import sqlite3
import pandas as pd
from tqdm import tqdm

# --- 1. Test Dataset (Example) ---
# Real evaluation mein aap apna test data validation loop mein dalenge
test_data = [
    {
        "schema": "CREATE TABLE users(id INT, name TEXT, age INT);",
        "question": "Find names of users older than 25",
        "target_sql": "SELECT name FROM users WHERE age > 25;"
    },
    {
        "schema": "CREATE TABLE employees(emp_id INT, salary INT, dept TEXT);",
        "question": "Show all departments with salary more than 50000",
        "target_sql": "SELECT dept FROM employees WHERE salary > 50000;"
    }
]

# --- 2. Evaluation Setup ---
correct_executions = 0
total_samples = len(test_data)

# Inference mode active karein
FastLanguageModel.for_inference(model)

print("Evaluating Model Accuracy...")

for sample in tqdm(test_data):
    # Temporary In-Memory Database setup karein data check karne ke liye
    conn = sqlite3.connect(":memory:")
    cursor = conn.cursor()

    try:
        # Schema execute karein
        cursor.execute(sample["schema"])
        conn.commit()

        # --- 3. Model se SQL Predict Karwayein ---
        formatted_prompt = prompt_template.format(
            schema=sample["schema"],
            question=sample["question"],
            sql=""
        )
        inputs = tokenizer([formatted_prompt], return_tensors="pt").to("cuda")

        outputs = model.generate(
            **inputs,
            max_new_tokens=50,
            temperature=0.1,
            do_sample=False,
            eos_token_id=tokenizer.eos_token_id
        )

        # Sirf predicted SQL extract karein
        input_len = inputs.input_ids.shape[1]
        predicted_sql = tokenizer.decode(outputs[0][input_len:], skip_special_tokens=True).strip()

        # --- 4. Execution Check (Dono queries run karke data match karein) ---
        # Target (Sahi) SQL ka output
        cursor.execute(sample["target_sql"])
        target_results = cursor.fetchall()

        # Model ke Predicted SQL ka output
        cursor.execute(predicted_sql)
        predicted_results = cursor.fetchall()

        # Agar dono database outputs exact same hain, toh prediction sahi hai
        if set(target_results) == set(predicted_results):
            correct_executions += 1

    except Exception as e:
        # Agar model ne invalid SQL banaya jo execute hi nahi ho paya
        pass
    finally:
        conn.close()

# --- 5. Final Accuracy Calculation ---
accuracy = (correct_executions / total_samples) * 100
print(f"\n🎯 Evaluation Results:")
print(f"Total Test Cases: {total_samples}")
print(f"Correctly Executed: {correct_executions}")
print(f"📊 Final Model Accuracy: {accuracy:.2f}%")
```

```python
import sqlite3
from tqdm import tqdm

# 1. Trained model ko inference (testing) mode par set karein
FastLanguageModel.for_inference(model)

# 2. Apna Test/Validation Dataset yahan dalein (Kuch examples niche hain)
test_dataset = [
    {
        "schema": "CREATE TABLE users(id INT, name TEXT, age INT);",
        "question": "Find names of users older than 25",
        "target_sql": "SELECT name FROM users WHERE age > 25;"
    },
    {
        "schema": "CREATE TABLE employees(emp_id INT, name TEXT, salary INT, dept TEXT);",
        "question": "Show names of employees with salary more than 50000",
        "target_sql": "SELECT name FROM employees WHERE salary > 50000;"
    },
    {
        "schema": "CREATE TABLE products(prod_id INT, item_name TEXT, stock INT);",
        "question": "Get items with zero stock",
        "target_sql": "SELECT item_name FROM products WHERE stock = 0;"
    }
]

# Evaluation counters
correct_predictions = 0
total_samples = len(test_dataset)

print(f"🔄 Evaluating trained model on {total_samples} test cases...\n")

# Loop through test data
for idx, sample in enumerate(tqdm(test_dataset)):

    # --- STEP A: Model Prediction ---
    formatted_prompt = prompt_template.format(
        schema=sample["schema"],
        question=sample["question"],
        sql="" # Model yahan se text generate karega
    )

    inputs = tokenizer([formatted_prompt], return_tensors="pt").to("cuda")

    outputs = model.generate(
        **inputs,
        max_new_tokens=50,
        temperature=0.1,
        do_sample=False,
        eos_token_id=tokenizer.eos_token_id
    )

    # Slicing: Sirf naya generated SQL extract karein
    input_length = inputs.input_ids.shape[1]
    predicted_sql = tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True).strip()

    # --- STEP B: Execution Accuracy Check (DB Simulation) ---
    conn = sqlite3.connect(":memory:") # Temporary Database
    cursor = conn.cursor()

    try:
        # DB mein table create karein
        cursor.execute(sample["schema"])
        conn.commit()

        # Sahi (Target) query chala kar result nikaalein
        cursor.execute(sample["target_sql"])
        target_results = cursor.fetchall()

        # Trained Model ki Predicted query chala kar result nikaalein
        cursor.execute(predicted_sql)
        predicted_results = cursor.fetchall()

        # Dono ka output check karein (Order change ho sakta hai isliye set() kiya hai)
        if set(target_results) == set(predicted_results):
            correct_predictions += 1

    except sqlite3.Error:
        # Agar model ne galat SQL banaya jo execute hi nahi hua, toh yeh fail count hoga
        pass
    finally:
        conn.close()

# --- STEP C: Final Accuracy Score ---
accuracy_percentage = (correct_predictions / total_samples) * 100

print("\n" + "="*40)
print(f"📊 TRAINED MODEL EVALUATION RESULTS")
print(f"✅ Total Test Cases   : {total_samples}")
print(f"🎯 Correct Predictions : {correct_predictions}")
print(f"🚀 FINAL ACCURACY SCORE: {accuracy_percentage:.2f}%")
```
