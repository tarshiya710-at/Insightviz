```python
def tokenize(text):

    pattern = r"""'[^']*'|"[^"]*"|[A-Za-z_][A-Za-z0-9_$]*|\d+(?:\.\d+)?|!=|<>|<=|>=|==|[^\s]"""

    return re.findall(pattern, str(text))
```

```python
PAD_TOKEN = "<pad>"
UNK_TOKEN = "<unk>"
BOS_TOKEN = "<bos>"
EOS_TOKEN = "<eos>"

PAD_IDX = 0
UNK_IDX = 1
BOS_IDX = 2
EOS_IDX = 3

counter = Counter()

for text in train_data["input"]:
    counter.update(tokenize(text))

for text in train_data["target"]:
    counter.update(tokenize(text))

vocab = {
    PAD_TOKEN: PAD_IDX,
    UNK_TOKEN: UNK_IDX,
    BOS_TOKEN: BOS_IDX,
    EOS_TOKEN: EOS_IDX
}

for token, frequency in counter.items():
    if frequency >= 1 and token not in vocab:
        vocab[token] = len(vocab)

id_to_token = {v: k for k, v in vocab.items()}

print("Vocabulary size:", len(vocab))
```

```python
MAX_INPUT_LEN = 384
MAX_OUTPUT_LEN = 160

def encode_text(text, max_len):

    tokens = tokenize(text)

    ids = [BOS_IDX]

    ids += [
        vocab.get(token, UNK_IDX)
        for token in tokens[:max_len - 2]
    ]

    ids.append(EOS_IDX)

    return ids
```

```python
train_inputs = [
    encode_text(x, MAX_INPUT_LEN)
    for x in train_data["input"]
]

train_targets = [
    encode_text(x, MAX_OUTPUT_LEN)
    for x in train_data["target"]
]

val_inputs = [
    encode_text(x, MAX_INPUT_LEN)
    for x in val_data["input"]
]

val_targets = [
    encode_text(x, MAX_OUTPUT_LEN)
    for x in val_data["target"]
]

test_inputs = [
    encode_text(x, MAX_INPUT_LEN)
    for x in test_df["input"]
]

test_targets = [
    encode_text(x, MAX_OUTPUT_LEN)
    for x in test_df["target"]
]
```

```python
class TextSQLDataset(Dataset):

    def __init__(self, inputs, targets):
        self.inputs = inputs
        self.targets = targets

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, index):
        return self.inputs[index], self.targets[index]
```

```python
def collate_fn(batch):

    inputs, targets = zip(*batch)

    max_input = max(len(x) for x in inputs)
    max_target = max(len(x) for x in targets)

    padded_inputs = []
    padded_targets = []

    for x in inputs:
        padded_inputs.append(
            x + [PAD_IDX] * (max_input - len(x))
        )

    for y in targets:
        padded_targets.append(
            y + [PAD_IDX] * (max_target - len(y))
        )

    return (
        torch.tensor(padded_inputs, dtype=torch.long),
        torch.tensor(padded_targets, dtype=torch.long)
    )
```

```python
BATCH_SIZE = 8

train_dataset = TextSQLDataset(
    train_inputs, train_targets
)

val_dataset = TextSQLDataset(
    val_inputs, val_targets
)

test_dataset = TextSQLDataset(
    test_inputs, test_targets
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
class PositionalEncoding(nn.Module):

    def __init__(self, d_model, max_len=1000):
        super().__init__()

        pe = torch.zeros(max_len, d_model)

        position = torch.arange(
            0, max_len, dtype=torch.float
        ).unsqueeze(1)

        div_term = torch.exp(
            torch.arange(0, d_model, 2).float()
            * (-math.log(10000.0) / d_model)
        )

        pe[:, 0::2] = torch.sin(position * div_term)

        pe[:, 1::2] = torch.cos(
            position * div_term[:pe[:, 1::2].shape[1]]
        )

        pe = pe.unsqueeze(0)

        self.register_buffer("pe", pe)

    def forward(self, x):

        return x + self.pe[:, :x.size(1)]
```

```python
class TextToSQLTransformer(nn.Module):

    def __init__(
        self,
        vocab_size,
        d_model=256,
        nhead=8,
        num_layers=3,
        dim_feedforward=512,
        dropout=0.2
    ):

        super().__init__()

        self.d_model = d_model

        self.embedding = nn.Embedding(
            vocab_size,
            d_model,
            padding_idx=PAD_IDX
        )

        self.positional_encoding = PositionalEncoding(
            d_model
        )

        self.transformer = nn.Transformer(
            d_model=d_model,
            nhead=nhead,
            num_encoder_layers=num_layers,
            num_decoder_layers=num_layers,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True
        )

        self.generator = nn.Linear(
            d_model,
            vocab_size
        )

    def forward(self, src, tgt):

        src_padding_mask = src == PAD_IDX
        tgt_padding_mask = tgt == PAD_IDX

        tgt_len = tgt.size(1)

        causal_mask = torch.triu(
            torch.ones(
                tgt_len,
                tgt_len,
                device=tgt.device,
                dtype=torch.bool
            ),
            diagonal=1
        )

        src_emb = self.embedding(src) * math.sqrt(self.d_model)
        tgt_emb = self.embedding(tgt) * math.sqrt(self.d_model)

        src_emb = self.positional_encoding(src_emb)
        tgt_emb = self.positional_encoding(tgt_emb)

        output = self.transformer(
            src_emb,
            tgt_emb,
            tgt_mask=causal_mask,
            src_key_padding_mask=src_padding_mask,
            tgt_key_padding_mask=tgt_padding_mask,
            memory_key_padding_mask=src_padding_mask
        )

        return self.generator(output)
```

```python
model = TextToSQLTransformer(
    vocab_size=len(vocab),
    d_model=256,
    nhead=8,
    num_layers=3,
    dim_feedforward=512,
    dropout=0.2
).to(device)

print(model)
```

```python
criterion = nn.CrossEntropyLoss(
    ignore_index=PAD_IDX
)
```

```python
LEARNING_RATE = 0.0001

optimizer = optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=0.01
)
```

```python
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=0.000001
)
```

```python
def train_one_epoch(model, loader):

    model.train()

    total_loss = 0
    total_tokens = 0

    for src, tgt in loader:

        src = src.to(device)
        tgt = tgt.to(device)

        tgt_input = tgt[:, :-1]
        tgt_output = tgt[:, 1:]

        optimizer.zero_grad()

        output = model(src, tgt_input)

        loss = criterion(
            output.reshape(-1, output.size(-1)),
            tgt_output.reshape(-1)
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        valid_tokens = (tgt_output != PAD_IDX).sum().item()

        total_loss += loss.item() * valid_tokens
        total_tokens += valid_tokens

    return total_loss / max(total_tokens, 1)
```

```python
def validate(model, loader):

    model.eval()

    total_loss = 0
    total_tokens = 0

    correct_tokens = 0

    with torch.no_grad():

        for src, tgt in loader:

            src = src.to(device)
            tgt = tgt.to(device)

            tgt_input = tgt[:, :-1]
            tgt_output = tgt[:, 1:]

            output = model(src, tgt_input)

            loss = criterion(
                output.reshape(-1, output.size(-1)),
                tgt_output.reshape(-1)
            )

            predictions = output.argmax(dim=-1)

            mask = tgt_output != PAD_IDX

            valid_tokens = mask.sum().item()

            correct_tokens += (
                (predictions == tgt_output) & mask
            ).sum().item()

            total_loss += loss.item() * valid_tokens
            total_tokens += valid_tokens

    avg_loss = total_loss / max(total_tokens, 1)

    accuracy = (
        100 * correct_tokens / max(total_tokens, 1)
    )

    return avg_loss, accuracy
```

```python
EPOCHS = 20
PATIENCE = 4

best_val_loss = float("inf")
counter = 0

train_losses = []
val_losses = []
val_accuracies = []

for epoch in range(EPOCHS):

    train_loss = train_one_epoch(
        model, train_loader
    )

    val_loss, val_accuracy = validate(
        model, val_loader
    )

    scheduler.step(val_loss)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    val_accuracies.append(val_accuracy)

    current_lr = optimizer.param_groups[0]["lr"]

    print(
        f"Epoch {epoch+1}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Token Accuracy: {val_accuracy:.2f}% | "
        f"LR: {current_lr:.7f}"
    )

    if val_loss < best_val_loss:

        best_val_loss = val_loss
        counter = 0

        torch.save(
            model.state_dict(),
            "best_model.pth"
        )

        print("Best model saved")

    else:

        counter += 1

        print("No improvement:", counter)

    if counter >= PATIENCE:

        print("Early stopping triggered")
        break
```

```python
def generate_sql(src_ids, max_len=MAX_OUTPUT_LEN):

    model.eval()

    src = torch.tensor(
        [src_ids],
        dtype=torch.long,
        device=device
    )

    src_padding_mask = src == PAD_IDX

    with torch.no_grad():

        src_emb = model.embedding(src) * math.sqrt(
            model.d_model
        )

        src_emb = model.positional_encoding(src_emb)

        memory = model.transformer.encoder(
            src_emb,
            src_key_padding_mask=src_padding_mask
        )

        generated = [BOS_IDX]

        for _ in range(max_len - 1):

            tgt = torch.tensor(
                [generated],
                dtype=torch.long,
                device=device
            )

            tgt_emb = model.embedding(tgt) * math.sqrt(
                model.d_model
            )

            tgt_emb = model.positional_encoding(tgt_emb)

            tgt_len = tgt.size(1)

            causal_mask = torch.triu(
                torch.ones(
                    tgt_len,
                    tgt_len,
                    dtype=torch.bool,
                    device=device
                ),
                diagonal=1
            )

            decoder_output = model.transformer.decoder(
                tgt_emb,
                memory,
                tgt_mask=causal_mask,
                memory_key_padding_mask=src_padding_mask
            )

            logits = model.generator(
                decoder_output[:, -1, :]
            )

            logits[:, PAD_IDX] = float("-inf")
            logits[:, BOS_IDX] = float("-inf")

            next_token = logits.argmax(dim=-1).item()

            generated.append(next_token)

            if next_token == EOS_IDX:
                break

    result = []

    for token_id in generated[1:]:

        if token_id == EOS_IDX:
            break

        if token_id not in [PAD_IDX, BOS_IDX]:
            result.append(id_to_token[token_id])

    return " ".join(result)
```

```python
def normalize_sql(sql):

    sql = re.sub(r"\s+", " ", str(sql).strip())

    sql = sql.rstrip(";").strip()

    return sql


def calculate_exact_match(data, inputs, limit=None):

    correct = 0
    total = min(len(data), limit) if limit else len(data)

    for i in range(total):

        predicted_sql = generate_sql(inputs[i])

        actual_sql = data.iloc[i]["target"]

        if normalize_sql(predicted_sql) == normalize_sql(actual_sql):
            correct += 1

    accuracy = 100 * correct / max(total, 1)

    return accuracy, correct, total
```

```python
def check_truncation(data):

    input_truncated = 0
    target_truncated = 0

    for _, row in data.iterrows():

        if len(tokenize(row["input"])) > MAX_INPUT_LEN - 2:
            input_truncated += 1

        if len(tokenize(row["target"])) > MAX_OUTPUT_LEN - 2:
            target_truncated += 1

    print(
        "Input truncation:",
        round(100 * input_truncated / len(data), 2),
        "%"
    )

    print(
        "Target truncation:",
        round(100 * target_truncated / len(data), 2),
        "%"
    )


check_truncation(train_data)
check_truncation(val_data)
```

```python
def target_unknown_percentage(data):

    total = 0
    unknown = 0

    for text in data["target"]:

        tokens = tokenize(text)

        total += len(tokens)

        unknown += sum(
            token not in vocab for token in tokens
        )

    return 100 * unknown / max(total, 1)


print(
    "Validation target UNK:",
    round(target_unknown_percentage(val_data), 2),
    "%"
)

print(
    "Test target UNK:",
    round(target_unknown_percentage(test_df), 2),
    "%"
)
```

```python
def normalize_sql(sql):

    tokens = tokenize(sql)

    if tokens and tokens[-1] == ";":
        tokens = tokens[:-1]

    return " ".join(tokens)
```
